- *¿Cuáles son las fases de la etapa de Análisis (Front-End) de un compilador?*  
**RESPUESTA:**

*   **Análisis Léxico (o Lexicográfico):** Lee el programa fuente carácter por carácter. Agrupa esos caracteres en palabras con significado llamadas **lexemas** y genera como salida un flujo de **tokens**. Además, en esta fase se eliminan los espacios en blanco, tabuladores y comentarios.
*   **Análisis Sintáctico (Parsing):** Toma el flujo de tokens y comprueba que estén ordenados según las **reglas gramaticales** del lenguaje. Si la estructura es correcta, construye una representación jerárquica llamada **Árbol Sintáctico** (o AST); si no, reporta los famosos errores de sintaxis.
*   **Análisis Semántico:** Revisa el árbol sintáctico para asegurar que el programa tenga **coherencia lógica**. Su tarea principal es el **chequeo de tipos** (verificar que no sumes un texto con un número, por ejemplo) y comprobar que las variables utilizadas hayan sido declaradas.
*   **Generación de Código Intermedio:** Traduce la estructura validada del programa a una representación simplificada de bajo nivel, que es independiente de cualquier arquitectura física (como el código de tres direcciones o cuádruplas).

- *¿Qué diferencia a un Lexema de un Token?*  
**RESPUESTA:**

*   **Lexema:** Es la **secuencia concreta de caracteres** (el texto real en el código) que coincide con la regla de un patrón. Es la "palabra" literal que el programador escribió.
    *   *Ejemplos:* `x`, `contador`, `3.1416`, `while`.
*   **Token:** Es la **categoría gramatical o etiqueta abstracta** que el compilador le asigna a un grupo de lexemas afines para que el analizador sintáctico pueda trabajar. Suele representarse como un nombre abstracto (a menudo en mayúsculas) y puede llevar un valor de atributo opcional.
    *   *Ejemplos:* `ID` (para variables), `NUMERO` (para constantes), o tokens de palabra clave como `WHILE`.
*   **La analogía definitiva para tu examen:** El **Token** es la categoría genérica, como la palabra *“Fruta”*. El **Lexema** es el elemento real y específico que encaja en esa categoría, como *“Manzana”* o *“Plátano”*.

- *¿Qué función cumple la Tabla de Símbolos durante la compilación?*  
**RESPUESTA:**

La **Tabla de Símbolos** es una estructura de datos auxiliar de alto rendimiento (generalmente una tabla hash) que sirve como almacén central de información durante la compilación. Sus funciones principales en el examen se resumen en tres puntos:

1.  **Registrar Identificadores y Atributos:** Almacena todos los nombres que aparecen en el código (variables, funciones, constantes, tipos, parámetros) junto con sus propiedades asociadas, tales como su **tipo de dato**, **tamaño**, **ámbito o *scope*** (si es global o local) y **dirección de memoria**.
2.  **Soportar las validaciones semánticas:** Funciona como el validador del contexto. Permite comprobar de forma rápida si un identificador se está utilizando correctamente (por ejemplo, si una variable fue declarada antes de usarse) o reportar un error de redeclaración si el programador define dos variables con el mismo nombre en el mismo ámbito.
3.  **Facilitar la Generación de Código:** El generador de código (Back-End) consulta la tabla de símbolos para saber en qué posición de memoria o registro del procesador se debe almacenar y recuperar el valor de cada variable al traducir el programa a código máquina.

## El "Monstruo Completo" (Enlace Estático)  
Por defecto, GCC usa ***enlace dinámico*** usando una estructura de datos tabular, la tabla **PLT** (*Procedure Linkage Table*) ya que el código usa `printf`, el programa necesita conectarse con la biblioteca estándar de C (`libc.so`). El enlazador no copia el código de `printf` dentro del archivo para no duplicar espacio en el disco; en su lugar, crea un "trampolín" o acceso directo en una sección llamada `<printf@plt>`.

Cada vez que se invoca a `printf`, realmente salta a este bloque plt, el cual averigua en qué parte de la memoria RAM del sistema operativo se encuentra la verdadera función `printf` y redirige el control allí en tiempo de ejecución.

Para ver cómo el enlazador inyecta literalmente miles y miles de líneas copiando físicamente el código de `printf` y todas sus dependencias dentro del ejecutable, se debe compilar de forma estática:

```bash
# Compilación estática

!objdump -d hola_estatico | wc -l

````python
import token
import tokenize
from io import BytesIO

# Código fuente de prueba en formato cadena
codigo_fuente = "suma = a + 10"

# Tokenización del código fuente
tokens = tokenize.tokenize(BytesIO(codigo_fuente.encode("utf-8")).readline)

print(
    f"{'LINEA/COL':<12} | {'TIPO DE TOKEN':<20} | {'VALOR (LEXEMA)':<15}"
)
print("-" * 55)
for tok in tokens:
    if tok.type in (
        tokenize.ENCODING,
        tokenize.ENDMARKER,
        tokenize.NL,
        tokenize.NEWLINE,
    ):
        continue
    nombre_token = token.tok_name[tok.type]
    posicion = f"{tok.start[0]}:{tok.start[1]}"
    print(f"{posicion:<12} | {nombre_token:<20} | {tok.string:<15}")
````

## 5. Sección Práctica 3: Inspección de la Tabla de Símbolos y Bytecode
En esta sección simularemos la interacción con la Tabla de Símbolos y observaremos la generación de código intermedio.

### [CÓDIGO 3.1] Simulación de una Tabla de Símbolos básica en Python
```python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )


# Prueba de la Tabla de Símbolos
ts = TablaDeSimbolos()
ts.insertar("a", "ENTERO", 5)
ts.insertar("b", "ENTERO", 10)
ts.insertar("suma", "ENTERO", 15)
ts.mostrar()
```


````python
class TablaDeSimbolos:

    def __init__(self):
        self.simbolos = {}

    def insertar(self, nombre, tipo, valor=None, ambito="global"):
        if nombre in self.simbolos:
            print(f"[ERROR LÉXICO/SINTÁCTICO] Identificador '{nombre}' ya declarado.")
        else:
            self.simbolos[nombre] = {
                "tipo": tipo,
                "valor": valor,
                "ambito": ambito,
            }
            print(f"[TABLA DE SÍMBOLOS] Insertado: {nombre} ({tipo})")

    def buscar(self, nombre):
        return self.simbolos.get(nombre, None)

    def mostrar(self):
        print("\n=== CONTENIDO DE LA TABLA DE SÍMBOLOS ===")
        print(f"{'NOMBRE':<12} | {'TIPO':<10} | {'ÁMBITO':<10} | {'VALOR':<10}")
        print("-" * 50)
        for nombre, datos in self.simbolos.items():
            print(
                f"{nombre:<12} | {datos['tipo']:<10} | {datos['ambito']:<10} | {str(datos['valor']):<10}"
            )


# Prueba de la Tabla de Símbolos
ts = TablaDeSimbolos()
ts.insertar("a", "ENTERO", 5)
ts.insertar("b", "ENTERO", 10)
ts.insertar("suma", "ENTERO", 15)
ts.mostrar()
````

Sí, existe una ambigüedad clásica: palabras reservadas como `FORWARD` y `ROTATE` cumplen perfectamente con el patrón general de un "Identificador".

**Cómo lo resolverá el analizador:**
1. **Prioridad en Expresiones Regulares:** Al construir el analizador léxico, las reglas de las palabras clave se evalúan antes que la regla genérica de identificadores.
2. **Validación o Diccionario de Búsqueda:** Otra técnica común es clasificar inicialmente todos los textos alfabéticos como identificadores y, antes de generar el token final, buscar el lexema en una tabla de palabras reservadas.

2. **Pregunta 2 (Estructura de la Tabla de Símbolos):** De los elementos de tu lenguaje original, ¿cuáles atributos (tipo, valor, ámbito, dirección de memoria) necesitará almacenar tu Tabla de Símbolos cuando se procese una declaración?

**RESPUESTA:**

Los siguientes atributos:

1. **Nombre (Identificador):** La clave única para buscar el símbolo en la tabla (ej. `x`).
2. **Tipo de Dato:** Necesario para validaciones semánticas (ej. saber si es un `Entero` para asegurar que las instrucciones como `FORWARD x` reciban un dato numérico y no texto).
3. **Valor actual:** Al ser un lenguaje de scripting o comandos interpretados sencillos, guardar el valor evaluado nos permite resolver referencias dinámicas en tiempo de ejecución de manera directa.
4. **Ámbito (Scope):** Aunque inicialmente es global, si el lenguaje crece para soportar subrutinas o bloques de comandos agrupados, es esencial registrar si la variable pertenece al ámbito global o es local a un bloque.
5. **Dirección de Memoria o Índice:** Para un DSL interpretado básico no es estrictamente obligatorio, pero si se compila a código intermedio o instrucciones de bajo nivel, se puede almacenar una dirección virtual o un índice de registro para facilitar la traducción de variables a memoria física.
